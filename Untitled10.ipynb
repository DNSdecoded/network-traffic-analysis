## Load the data

### Subtask:
Load the `isp.csv` file into a pandas DataFrame.


**Reasoning**:
Import the pandas library and load the data from the isp.csv file into a DataFrame.



```python
import pandas as pd

df = pd.read_csv('isp.csv')
display(df.head())
```

## Data cleaning and preparation

### Subtask:
Parse 'Login Time' to datetime objects, convert 'Session Time' from seconds to minutes, ensure 'Download' and 'Upload' are numeric types, create new columns for login day of the week and hour of the day, and calculate cumulative download and upload.


**Reasoning**:
Convert 'Login Time' to datetime, 'Session Time' to minutes, 'Download' and 'Upload' to numeric, create new time-based features, and calculate cumulative usage.



```python
df['Login Time'] = pd.to_datetime(df['Login Time'], format='%d-%m-%y %H:%M')

# Drop rows with NaN in 'Session Time' or where key columns are all NaN
df.dropna(subset=['Session Time'], inplace=True)
df.dropna(subset=['MAC Id', 'Login Time', 'Session Time', 'Download', 'Upload'], how='all', inplace=True)

def time_to_minutes(time_str):
    if pd.isna(time_str):
        return None
    parts = str(time_str).split(':')
    if len(parts) == 3:
        h, m, s = map(float, parts)
        return h * 60 + m + s / 60
    elif len(parts) == 2:
         m, s = map(float, parts)
         return m + s / 60
    return None

df['Session Time'] = df['Session Time'].apply(time_to_minutes)


df['Download'] = pd.to_numeric(df['Download'], errors='coerce').fillna(0)
df['Upload'] = pd.to_numeric(df['Upload'], errors='coerce').fillna(0)

df['login_day_of_week'] = df['Login Time'].dt.day_name()
df['login_hour'] = df['Login Time'].dt.hour

df['cumulative_download'] = df['Download'].cumsum()
df['cumulative_upload'] = df['Upload'].cumsum()

display(df.head())
display(df.info())
```

## Generate overall insights

### Subtask:
Calculate total download and upload in GB, find the total number of sessions, calculate the average session duration in minutes, determine the top 5 peak usage days for download and upload, calculate the upload vs download ratio, and identify the longest and shortest sessions.


**Reasoning**:
Calculate the required metrics and find the top 5 sessions and days as per the instructions.



```python
total_download_gb = df['Download'].sum() / 1024
total_upload_gb = df['Upload'].sum() / 1024

total_sessions = len(df)

average_session_time_minutes = df['Session Time'].mean()

df['Login_Date'] = df['Login Time'].dt.date
daily_usage = df.groupby('Login_Date')[['Download', 'Upload']].sum()
top_5_download_days = daily_usage.sort_values(by='Download', ascending=False).head(5)
top_5_upload_days = daily_usage.sort_values(by='Upload', ascending=False).head(5)

upload_download_ratio = total_upload_gb / total_download_gb if total_download_gb > 0 else 0

longest_session = df.loc[df['Session Time'].idxmax()]
shortest_session = df.loc[df['Session Time'].idxmin()]

overall_metrics = {
    'Total Sessions': total_sessions,
    'Total Download (GB)': total_download_gb,
    'Total Upload (GB)': total_upload_gb,
    'Average Session Time (minutes)': average_session_time_minutes
}

display("Overall Metrics:")
display(overall_metrics)
display("\nTop 5 Download Days:")
display(top_5_download_days)
display("\nTop 5 Upload Days:")
display(top_5_upload_days)
display(f"\nUpload to Download Ratio: {upload_download_ratio:.2f}")
display("\nLongest Session:")
display(longest_session)
display("\nShortest Session:")
display(shortest_session)
```

Aggregate the data by day, hour, day of the week, and day of the week and hour combinations as requested in the instructions.

Analyze the internet usage data from the file "isp.csv", focusing on session behavior, peak usage times, weekday vs weekend usage, upload vs download ratio, anomaly detection, most active day, and daily session duration trends.
```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.ticker import MultipleLocator

# Reload the data to ensure a clean start
df = pd.read_csv('isp.csv')

# Convert 'Login Time' to datetime
df['Login Time'] = pd.to_datetime(df['Login Time'], format='%d-%m-%y %H:%M')

# Drop rows with NaN in 'Session Time' or where key columns are all NaN
df.dropna(subset=['Session Time'], inplace=True)
df.dropna(subset=['MAC Id', 'Login Time', 'Session Time', 'Download', 'Upload'], how='all', inplace=True)


def time_to_minutes(time_str):
    if pd.isna(time_str):
        return None
    parts = str(time_str).split(':')
    if len(parts) == 3:
        h, m, s = map(float, parts)
        return h * 60 + m + s / 60
    elif len(parts) == 2:
        m, s = map(float, parts)
        return m + s / 60
    return None

df['Session Time'] = df['Session Time'].apply(time_to_minutes)

# Drop rows where 'Session Time' is NaN after conversion
df.dropna(subset=['Session Time'], inplace=True)

df['Download'] = pd.to_numeric(df['Download'], errors='coerce').fillna(0)
df['Upload'] = pd.to_numeric(df['Upload'], errors='coerce').fillna(0)

df['login_day_of_week'] = df['Login Time'].dt.day_name()
df['login_hour'] = df['Login Time'].dt.hour

df['cumulative_download'] = df['Download'].cumsum()
df['cumulative_upload'] = df['Upload'].cumsum()

# Display DataFrame and info after cleaning for debugging
display(df.head())
display(df.info())


# Total number of sessions
total_sessions = len(df)
print(f"Total number of sessions: {total_sessions}")

# Average session time
average_session_time = df['Session Time'].mean()
print(f"Average session time: {average_session_time:.2f} minutes")

# Longest and shortest sessions
longest_session = df.loc[df['Session Time'].idxmax()]
shortest_session = df.loc[df['Session Time'].idxmin()]

print("\nLongest Session:")
display(longest_session)

print("\nShortest Session:")
display(shortest_session)

# Sessions per day
sessions_per_day = df.groupby(df['Login Time'].dt.date).size()
print("\nSessions per day:")
display(sessions_per_day)

# Plot sessions per day
plt.figure(figsize=(12, 6))
sessions_per_day.plot(kind='bar')
plt.title('Number of Sessions Per Day')
plt.xlabel('Date')
plt.ylabel('Number of Sessions')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Analyze peak usage time (Hour of Day)
df['login_hour'] = df['Login Time'].dt.hour

# Heatmap of hourly login distribution across days
login_distribution_by_day_hour = df.groupby(['login_day_of_week', 'login_hour']).size().unstack(fill_value=0)

plt.figure(figsize=(12, 8))
sns.heatmap(login_distribution_by_day_hour, annot=True, fmt='d', cmap='YlGnBu')
plt.xlabel('Hour of Day')
plt.ylabel('Day of Week')
plt.title('Hourly Login Distribution Across Days')
plt.tight_layout()
plt.show()

# Average usage by day and hour
average_usage_by_day_hour = df.groupby(['login_day_of_week', 'login_hour'])[['Download', 'Upload', 'Session Time']].mean()
display("Average usage by day and hour:")
display(average_usage_by_day_hour)
```

```python
# Classify days as weekday or weekend
df['day_type'] = df['Login Time'].dt.dayofweek.apply(lambda x: 'Weekend' if x >= 5 else 'Weekday')

# Compare data usage on weekdays vs weekends
weekday_weekend_usage = df.groupby('day_type')[['Download', 'Upload']].sum().reset_index()

# Plot weekday vs weekend usage
plt.figure(figsize=(8, 6))
sns.barplot(x='day_type', y='Download', data=weekday_weekend_usage, color='skyblue', label='Download')
sns.barplot(x='day_type', y='Upload', data=weekday_weekend_usage, color='lightcoral', bottom=weekday_weekend_usage['Download'], label='Upload')
plt.xlabel('Day Type')
plt.ylabel('Total Data Usage (MB)')
plt.title('Weekday vs Weekend Data Usage')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Upload vs Download Ratio
total_download = df['Download'].sum()
total_upload = df['Upload'].sum()
total_data = total_download + total_upload

upload_download_ratio = total_upload / total_download if total_download > 0 else 0

print(f"\nUpload to Download Ratio: {upload_download_ratio:.2f}")

# Plot upload vs download data share (Pie chart)
labels = ['Upload', 'Download']
sizes = [total_upload, total_download]
colors = ['lightcoral', 'lightskyblue']
plt.figure(figsize=(8, 8))
plt.pie(sizes, labels=labels, colors=colors, autopct='%1.1f%%', startangle=140)
plt.title('Upload vs Download Data Share')
plt.axis('equal')
plt.show()
```

```python
# Anomaly Detection

# Define thresholds (these can be adjusted based on your data and domain knowledge)
download_threshold = df['Download'].quantile(0.99) # sessions with download in the top 1%
upload_threshold = df['Upload'].quantile(0.99)     # sessions with upload in the top 1%
session_time_threshold_long = df['Session Time'].quantile(0.99) # sessions in the top 1% duration
session_time_threshold_short = 1 # sessions shorter than 1 minute (rapid relogins)
rapid_relogin_threshold_seconds = 60 # sessions with login times less than 60 seconds apart

# Identify anomalous sessions based on data usage and session time
anomalous_sessions_download = df[df['Download'] > download_threshold]
anomalous_sessions_upload = df[df['Upload'] > upload_threshold]
anomalous_sessions_duration = df[df['Session Time'] > session_time_threshold_long]
rapid_relogin_sessions = df[df['Session Time'] < session_time_threshold_short]


print("\nAnomalous Sessions (High Download):")
display(anomalous_sessions_download)

print("\nAnomalous Sessions (High Upload):")
display(anomalous_sessions_upload)

print("\nAnomalous Sessions (Long Duration):")
display(anomalous_sessions_duration)

print("\nRapid Relogin Sessions (Very Short Duration):")
display(rapid_relogin_sessions)

# Optional: Further investigate rapid relogins by looking at time difference between consecutive sessions for each MAC ID
df['time_diff'] = df.groupby('MAC Id')['Login Time'].diff().dt.total_seconds().fillna(0)
rapid_relogin_sessions_by_time = df[df['time_diff'] < rapid_relogin_threshold_seconds]

print("\nRapid Relogin Sessions (Based on Login Time Difference):")
display(rapid_relogin_sessions_by_time)
```

```python
# Daily Session Duration Trend
daily_session_time = df.groupby(df['Login Time'].dt.date)['Session Time'].sum()

plt.figure(figsize=(12, 6))
daily_session_time.plot(kind='line', marker='o')
plt.title('Total Session Time Per Day')
plt.xlabel('Date')
plt.ylabel('Total Session Time (minutes)')
plt.xticks(rotation=45)
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```
